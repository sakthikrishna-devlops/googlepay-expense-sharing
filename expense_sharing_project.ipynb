```python

import numpy as np
from prettytable import PrettyTable

friends = ["Alice", "Bob", "Carol", "David"]
expense_matrix = np.zeros((len(friends), len(friends)))
transaction_history = []

# Add and split expenses
def add_expense(payer, beneficiaries, amount, description, split_type="equal", custom_shares=None):
    payer_idx = friends.index(payer)

    if split_type == "equal":
        share_per_person = amount / len(beneficiaries)

        for beneficiary in beneficiaries:
            beneficiary_idx = friends.index(beneficiary)
            expense_matrix[payer_idx][beneficiary_idx] += share_per_person

    elif split_type == "custom":
        for beneficiary in beneficiaries:
            beneficiary_idx = friends.index(beneficiary)
            share = custom_shares[beneficiary]
            expense_matrix[payer_idx][beneficiary_idx] += share

    transaction_history.append(
        [payer, description, amount, beneficiaries.copy()]
    )

    print("\nExpense added successfully!")
    print("Description:", description)
    print("Amount: ₹", amount)

# Display expense history
def display_history():
    table = PrettyTable()
    table.field_names = ["Payer", "Description", "Amount", "Beneficiaries"]

    for transaction in transaction_history:
        table.add_row([
            transaction[0],
            transaction[1],
            f"₹{transaction[2]:.2f}",
            ", ".join(transaction[3])
        ])

    print("\nTransaction History:")
    print(table)

# Display each user's transaction history
def user_history():
    for friend in friends:
        print("\nTransaction history for", friend)

        for transaction in transaction_history:
            if friend == transaction[0] or friend in transaction[3]:
                print(
                    transaction[1],
                    "- ₹",
                    f"{transaction[2]:.2f}",
                    "- Paid by:",
                    transaction[0]
                )

# Calculate balances
def calculate_settlements():
    total_paid = np.sum(expense_matrix, axis=1)
    total_owed = np.sum(expense_matrix, axis=0)
    net_balance = total_paid - total_owed
    return net_balance

# Display balance summary
def display_settlements():
    settlements = calculate_settlements()

    table = PrettyTable()
    table.field_names = ["Friend", "Settlement"]

    for i, friend in enumerate(friends):
        if settlements[i] > 0:
            table.add_row([
                friend, f"should receive ₹{settlements[i]:.2f}"
            ])
        elif settlements[i] < 0:
            table.add_row([
                friend, f"owes ₹{-settlements[i]:.2f}"
            ])
        else:
            table.add_row([friend, "is settled"])

    print("\nFinal settlements:")
    print(table)

# Display shared amounts for an expense
def display_shared_amounts():
    table = PrettyTable()
    table.field_names = ["Payer", "Beneficiary", "Shared Amount"]

    for i in range(len(friends)):
        for j in range(len(friends)):
            if expense_matrix[i][j] > 0:
                table.add_row([
                    friends[i],
                    friends[j],
                    f"₹{expense_matrix[i][j]:.2f}"
                ])

    print("\nShared Expense Details:")
    print(table)

# Input expenses
add_expense(
    "Alice",
    ["Alice", "Bob", "Carol"],
    1200,
    "Dinner"
)

add_expense(
    "Bob",
    ["Bob", "Carol", "David"],
    900,
    "Movie tickets"
)

add_expense(
    "Carol",
    ["Alice", "Bob", "Carol", "David"],
    1600,
    "Groceries"
)

# Custom split example
add_expense(
    "David",
    ["Alice", "David"],
    1000,
    "Shopping",
    "custom",
    {"Alice": 700, "David": 300}
)

# Display results
display_shared_amounts()
display_history()
user_history()
display_settlements()
```
